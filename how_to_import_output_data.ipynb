# How to import Data Outputs 

When working with the output CSV files, note that leading zeros in some columns (e.g., StateFIPS, LEADID, CensusTract, County) may be dropped when the data is re-imported due to how Pandas and CSV files handle numerical data. To ensure consistency and preserve leading zeros, you must use the pad_column_values function.

```python
import os
import chardet
import pandas as pd

from my_functions import pad_column_values
```

```python
current_directory = os.getcwd()
print(current_directory)
```

### Non-Aggregated Data Outputs 
*Note, extremely large when a single file. Consequently, import using parts and using chunks*

```python
# Directory containing the split files
input_dir = "data_outputs/internet_access_nonaggregated"

# List all CSV files in the directory
file_paths = [os.path.join(input_dir, file) for file in os.listdir(input_dir) if file.endswith('.csv')]

# Read and concatenate the files
chunks = []
try:
    for file_path in file_paths:
        for chunk in pd.read_csv(file_path, chunksize=100000, encoding='ISO-8859-1', on_bad_lines='skip'):
            chunks.append(chunk)

    # Combine all chunks into a single DataFrame
    data = pd.concat(chunks, ignore_index=True)
    print("Files combined and loaded successfully.")
except UnicodeDecodeError as e:
    print(f"UnicodeDecodeError: {e}")
except Exception as e:
    print(f"An error occurred: {e}")

# Pad columns
data = pad_column_values(data, "StateFIPS", 2)
data = pad_column_values(data, "County", 5)
data = pad_column_values(data, "CensusTract", 11)
data = pad_column_values(data, "CensusBlock", 4)
data = pad_column_values(data, "LEADID", 7)

# View data
print(data.head(5))
```
